```python
#Question 1
class SmartLight:
    def __init__(self, name):
        self.name = name

    def set_status(self, action):
        print(f"{self.name} is {action}")


# Sample Input
light_name = "Bedroom Light"
action = "ON"

# Object creation
light = SmartLight(light_name)
light.set_status(action)
```

```python
#Question 2
class Employee:
    def __init__(self, name, emp_id, department):
        self.name = name
        self.emp_id = emp_id
        self.department = department

    def display_id_card(self):
        print("Employee ID Card")
        print("Name:", self.name)
        print("ID:", self.emp_id)
        print("Department:", self.department)


# Sample Input
emp = Employee("Rahul", "EMP102", "IT")
emp.display_id_card()
```

```python
#Question 3
class Contact:
    def __init__(self, name, phone):
        self.name = name
        self.phone = phone

    def display_contact(self):
        print("Contact Saved")
        print("Name:", self.name)
        print("Phone:", self.phone)


# Sample Input
contact = Contact("Anita", "9876543210")
contact.display_contact()
```

```python
#Question 4
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def display_price_tag(self):
        print("Product:", self.name)
        print(f"Price: ₹{self.price}")


# Sample Input
product = Product("Headphones", 2499)
product.display_price_tag()
```

```python
#Question 5
class Movie:
    def __init__(self, name, rating):
        self.name = name
        self.rating = rating

    def display_movie(self):
        print("Movie:", self.name)
        print(f"Rating: {self.rating} / 5")


# Sample Input
movie = Movie("Inception", 4.8)
movie.display_movie()
```

```python
#Question 6
class Delivery:
    def __init__(self, customer, address):
        self.customer = customer
        self.address = address

    def display_delivery_details(self):
        print("Delivery Details")
        print("Customer:", self.customer)
        print("Address:", self.address)


# Sample Input
delivery = Delivery("Suman", "Hyderabad")
delivery.display_delivery_details()
```
